```python

# TASK 3: MISSING VALUE ANALYSIS

print("=" * 60)
print("TASK 3: MISSING VALUE ANALYSIS")
print("=" * 60)

# 1. Standardize column names
df.columns = df.columns.str.strip()

# 2. Clean whitespace in categorical columns
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()

# 3. Convert common empty representations into NaN
df.replace(
    ["", " ", "NA", "N/A", "NULL", "null"],
    np.nan,
    inplace=True
)

# 4. Calculate missing values
missing_analysis = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().mean() * 100).round(2),
    "Available Count": df.notnull().sum()
})

# 5. Sort by missing percentage
missing_analysis = missing_analysis.sort_values(
    by="Missing Percentage",
    ascending=False
)

print("\nMISSING VALUE REPORT")
display(missing_analysis)

# 6. Total missing values
total_missing = df.isnull().sum().sum()

print("\nTOTAL MISSING VALUES:", total_missing)

print("\nTOTAL AVAILABLE VALUES:", df.notnull().sum().sum())

# 7. Visualize missing values
plt.figure(figsize=(16, 7))

missing_analysis["Missing Percentage"].plot(
    kind="bar"
)

plt.title("Missing Values Analysis - NCBI ClinVar Dataset")
plt.xlabel("Dataset Columns")
plt.ylabel("Missing Percentage (%)")
plt.xticks(rotation=90)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()

# 8. Display columns with highest missing values
print("\nTOP 10 COLUMNS WITH MISSING VALUES")

display(missing_analysis.head(10))

print("\nTASK 3 COMPLETED SUCCESSFULLY")
```

```python
# TASK 8: CORRELATION ANALYSIS

print("=" * 60)
print("TASK 8: CORRELATION ANALYSIS")
print("=" * 60)

# 1. Select numerical columns
numerical_df = df.select_dtypes(include=["int64", "float64"])

print("\n1. NUMERICAL FEATURES")
print(numerical_df.columns.tolist())

# 2. Calculate correlation matrix
correlation_matrix = numerical_df.corr()

print("\n2. CORRELATION MATRIX")
display(correlation_matrix.round(2))

# 3. Visualize correlation matrix
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Matrix of Numerical Features")
plt.tight_layout()
plt.show()

# 4. Identify highly correlated feature pairs
print("\n3. HIGHLY CORRELATED FEATURE PAIRS")

corr_pairs = correlation_matrix.abs().unstack()

corr_pairs = corr_pairs[
    (corr_pairs < 1.0) & (corr_pairs >= 0.80)
]

corr_pairs = corr_pairs.drop_duplicates()

if len(corr_pairs) > 0:
    display(corr_pairs.to_frame(name="Correlation"))
else:
    print("No highly correlated feature pairs found.")

print("\nTASK 8 COMPLETED SUCCESSFULLY")
```

```python
# TASK 9: OUTLIER DETECTION USING IQR METHOD

print("=" * 60)
print("TASK 9: OUTLIER DETECTION")
print("=" * 60)

# 1. Select numerical columns
numerical_columns = [
    "POS", "AF_ESP", "AF_EXAC", "AF_TGP",
    "LoFtool", "CADD_PHRED", "CADD_RAW", "BLOSUM62"
]

print("\n1. OUTLIER DETECTION USING IQR")

outlier_summary = []

for col in numerical_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    outlier_summary.append({
        "Feature": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Outlier Count": len(outliers)
    })

outlier_df = pd.DataFrame(outlier_summary)

display(outlier_df.round(4))

# 2. Visualize outliers using boxplots
print("\n2. BOXPLOT VISUALIZATION")

for col in numerical_columns:

    plt.figure(figsize=(8, 3))

    sns.boxplot(x=df[col])

    plt.title("Outlier Detection - " + col)
    plt.xlabel(col)

    plt.tight_layout()
    plt.show()

print("\nTASK 9 COMPLETED SUCCESSFULLY")
```

```python
# TASK 10: DATA CLEANING AND PREPARATION

print("=" * 60)
print("TASK 10: DATA CLEANING AND PREPARATION")
print("=" * 60)

# 1. Create a backup of the original dataset
df_clean = df.copy()

print("\n1. ORIGINAL DATASET")
print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])

# 2. Remove unnecessary spaces from column names
df_clean.columns = df_clean.columns.str.strip()

# 3. Remove leading and trailing spaces from text values
for col in df_clean.select_dtypes(include=["object"]).columns:
    df_clean[col] = df_clean[col].str.strip()

# 4. Replace empty strings with missing values
df_clean.replace(r"^\s*$", np.nan, regex=True, inplace=True)

# 5. Check missing values
print("\n2. MISSING VALUES AFTER CLEANING")

missing_values = df_clean.isnull().sum()

display(
    missing_values[missing_values > 0]
    .to_frame(name="Missing Count")
    .sort_values(by="Missing Count", ascending=False)
)

# 6. Check duplicate records
print("\n3. DUPLICATE RECORD CHECK")

print("Duplicate Rows:", df_clean.duplicated().sum())

# 7. Verify target column
print("\n4. TARGET COLUMN VALIDATION")

print("CLASS unique values:", sorted(df_clean["CLASS"].dropna().unique()))
print("Missing CLASS values:", df_clean["CLASS"].isnull().sum())

# 8. Final dataset information
print("\n5. CLEANED DATASET INFORMATION")

print("Dataset Shape:", df_clean.shape)
print("\nColumn Data Types:")
print(df_clean.dtypes.value_counts())

print("\nTASK 10 COMPLETED SUCCESSFULLY")
```

```python
# TASK 11: HANDLING MISSING VALUES

print("=" * 60)
print("TASK 11: HANDLING MISSING VALUES")
print("=" * 60)

# 1. Create a copy
df_processed = df_clean.copy()

# 2. Identify extremely sparse columns
sparse_columns = [
    "MOTIF_NAME",
    "MOTIF_POS",
    "MOTIF_SCORE_CHANGE",
    "HIGH_INF_POS",
    "DISTANCE",
    "SSR",
    "CLNDNINCL",
    "CLNSIGINCL",
    "CLNDISDBINCL"
]

print("\n1. REMOVING EXTREMELY SPARSE COLUMNS")

df_processed.drop(
    columns=sparse_columns,
    inplace=True,
    errors="ignore"
)

print("Removed Columns:", sparse_columns)
print("Shape After Column Removal:", df_processed.shape)

# 3. Separate numerical and categorical columns
numerical_columns = df_processed.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_columns = df_processed.select_dtypes(
    include=["object"]
).columns.tolist()

# Do not impute the target column
if "CLASS" in numerical_columns:
    numerical_columns.remove("CLASS")

print("\n2. NUMERICAL MISSING VALUE IMPUTATION")

for col in numerical_columns:
    median_value = df_processed[col].median()
    df_processed[col] = df_processed[col].fillna(median_value)

print("Numerical missing values filled using median.")

# 4. Handle categorical missing values
print("\n3. CATEGORICAL MISSING VALUE IMPUTATION")

for col in categorical_columns:
    df_processed[col] = df_processed[col].fillna("Unknown")

print("Categorical missing values filled using Unknown.")

# 5. Verify missing values
print("\n4. MISSING VALUE VERIFICATION")

remaining_missing = df_processed.isnull().sum().sum()

print("Total Missing Values:", remaining_missing)

# 6. Final dataset information
print("\n5. FINAL PROCESSED DATASET")

print("Rows:", df_processed.shape[0])
print("Columns:", df_processed.shape[1])

print("\nRemaining Missing Values by Column:")
display(
    df_processed.isnull().sum()
    .to_frame(name="Missing Count")
    .query("`Missing Count` > 0")
)

print("\nTASK 11 COMPLETED SUCCESSFULLY")
```

```python
# TASK 12: FEATURE SELECTION

print("=" * 60)
print("TASK 12: FEATURE SELECTION")
print("=" * 60)

# 1. Create input features and target variable
print("\n1. SEPARATING FEATURES AND TARGET")

X = df_processed.drop(columns=["CLASS"])
y = df_processed["CLASS"]

print("Input Features Shape:", X.shape)
print("Target Variable Shape:", y.shape)

# 2. Display input feature columns
print("\n2. INPUT FEATURE COLUMNS")

print("Total Input Features:", len(X.columns))

for i, col in enumerate(X.columns, start=1):
    print(i, ".", col)

# 3. Display target variable distribution
print("\n3. TARGET VARIABLE DISTRIBUTION")

display(
    y.value_counts()
    .sort_index()
    .to_frame(name="Record Count")
)

# 4. Calculate target class percentages
print("\n4. TARGET CLASS PERCENTAGES")

class_percentages = (
    y.value_counts(normalize=True)
    .sort_index() * 100
)

display(
    class_percentages.to_frame(name="Percentage")
)

# 5. Verify feature and target separation
print("\n5. FEATURE AND TARGET VALIDATION")

print("CLASS present in X:", "CLASS" in X.columns)
print("CLASS present in y:", y.name == "CLASS")

print("\nFinal Feature Shape:", X.shape)
print("Final Target Shape:", y.shape)

print("\nTASK 12 COMPLETED SUCCESSFULLY")
```

```python
# TASK 13: TRAIN-TEST SPLIT

print("=" * 60)
print("TASK 13: TRAIN-TEST SPLIT")
print("=" * 60)

from sklearn.model_selection import train_test_split

# 1. Split the dataset
print("\n1. SPLITTING DATASET")

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# 2. Display training and testing shapes
print("\n2. DATASET SHAPES")

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)

# 3. Training class distribution
print("\n3. TRAINING CLASS DISTRIBUTION")

display(
    y_train.value_counts()
    .sort_index()
    .to_frame(name="Record Count")
)

# 4. Testing class distribution
print("\n4. TESTING CLASS DISTRIBUTION")

display(
    y_test.value_counts()
    .sort_index()
    .to_frame(name="Record Count")
)

# 5. Verify class proportions
print("\n5. CLASS DISTRIBUTION PERCENTAGES")

print("Original Dataset:")
display((y.value_counts(normalize=True) * 100).sort_index())

print("Training Dataset:")
display((y_train.value_counts(normalize=True) * 100).sort_index())

print("Testing Dataset:")
display((y_test.value_counts(normalize=True) * 100).sort_index())

print("\nTASK 13 COMPLETED SUCCESSFULLY")
```

```python
# TASK 14: FEATURE TYPE IDENTIFICATION

print("=" * 60)
print("TASK 14: FEATURE TYPE IDENTIFICATION")
print("=" * 60)

# 1. Identify numerical and categorical features
print("\n1. IDENTIFYING FEATURE TYPES")

numerical_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object"]
).columns.tolist()

print("Numerical Features:", len(numerical_features))
print(numerical_features)

print("\nCategorical Features:", len(categorical_features))
print(categorical_features)

# 2. Display numerical feature summary
print("\n2. NUMERICAL FEATURE SUMMARY")

display(X_train[numerical_features].describe().T)

# 3. Identify categorical feature cardinality
print("\n3. CATEGORICAL FEATURE CARDINALITY")

cardinality = pd.DataFrame({
    "Feature": categorical_features,
    "Unique Values": [
        X_train[col].nunique() for col in categorical_features
    ]
})

display(
    cardinality.sort_values(
        by="Unique Values",
        ascending=False
    )
)

# 4. Identify high-cardinality features
print("\n4. HIGH-CARDINALITY FEATURES")

high_cardinality = cardinality[
    cardinality["Unique Values"] > 50
]

display(high_cardinality)

# 5. Final feature type summary
print("\n5. FEATURE TYPE SUMMARY")

print("Total Features:", X_train.shape[1])
print("Numerical Features:", len(numerical_features))
print("Categorical Features:", len(categorical_features))

print("\nTASK 14 COMPLETED SUCCESSFULLY")
```

```python
# TASK 15: REMOVE IDENTIFIER AND POTENTIAL LEAKAGE FEATURES

print("=" * 60)
print("TASK 15: REMOVE IDENTIFIER AND POTENTIAL LEAKAGE FEATURES")
print("=" * 60)

# 1. Create a copy of training and testing features
X_train_selected = X_train.copy()
X_test_selected = X_test.copy()

# 2. Identify columns to exclude
columns_to_remove = [
    "CLNDISDB",
    "CLNDN",
    "CLNHGVS",
    "CLNVI",
    "Feature"
]

print("\n1. COLUMNS SELECTED FOR REMOVAL")

for col in columns_to_remove:
    if col in X_train_selected.columns:
        print(col)

# 3. Remove selected columns
X_train_selected.drop(
    columns=columns_to_remove,
    inplace=True,
    errors="ignore"
)

X_test_selected.drop(
    columns=columns_to_remove,
    inplace=True,
    errors="ignore"
)

# 4. Display updated dataset shapes
print("\n2. UPDATED DATASET SHAPES")

print("X_train_selected:", X_train_selected.shape)
print("X_test_selected:", X_test_selected.shape)

# 5. Verify removed columns
print("\n3. REMOVAL VALIDATION")

remaining_columns = [
    col for col in columns_to_remove
    if col in X_train_selected.columns
]

print("Removed Columns Remaining:", remaining_columns)

# 6. Display remaining features
print("\n4. REMAINING FEATURES")

print("Total Features:", len(X_train_selected.columns))

for i, col in enumerate(X_train_selected.columns, start=1):
    print(i, ".", col)

print("\nTASK 15 COMPLETED SUCCESSFULLY")
```

```python
# TASK 16: CATEGORICAL AND NUMERICAL FEATURE SEPARATION

print("=" * 60)
print("TASK 16: CATEGORICAL AND NUMERICAL FEATURE SEPARATION")
print("=" * 60)

# 1. Identify numerical features
print("\n1. IDENTIFYING NUMERICAL FEATURES")

numerical_features = X_train_selected.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

# 2. Identify categorical features
print("\n2. IDENTIFYING CATEGORICAL FEATURES")

categorical_features = X_train_selected.select_dtypes(
    include=["object"]
).columns.tolist()

# 3. Move encoded biological attributes into categorical features
special_categorical = ["ORIGIN", "STRAND"]

for col in special_categorical:
    if col in numerical_features:
        numerical_features.remove(col)
        categorical_features.append(col)

# 4. Display numerical features
print("\n3. NUMERICAL FEATURES")

print("Total Numerical Features:", len(numerical_features))

for i, col in enumerate(numerical_features, start=1):
    print(i, ".", col)

# 5. Display categorical features
print("\n4. CATEGORICAL FEATURES")

print("Total Categorical Features:", len(categorical_features))

for i, col in enumerate(categorical_features, start=1):
    print(i, ".", col)

# 6. Verify feature separation
print("\n5. FEATURE SEPARATION VALIDATION")

print("Total Selected Features:", X_train_selected.shape[1])

print(
    "Total Separated Features:",
    len(numerical_features) + len(categorical_features)
)

print(
    "All Features Identified:",
    set(numerical_features + categorical_features)
    == set(X_train_selected.columns)
)

print("\nTASK 16 COMPLETED SUCCESSFULLY")
```

```python
# TASK 17: CATEGORICAL FEATURE ENCODING

print("=" * 60)
print("TASK 17: CATEGORICAL FEATURE ENCODING")
print("=" * 60)

# 1. Create copies of training and testing data
X_train_encoded = X_train_selected.copy()
X_test_encoded = X_test_selected.copy()

# 2. Identify low and high cardinality categorical features
print("\n1. IDENTIFYING CATEGORICAL FEATURES")

low_cardinality = [
    col for col in categorical_features
    if X_train_encoded[col].nunique() <= 10
]

high_cardinality = [
    col for col in categorical_features
    if X_train_encoded[col].nunique() > 10
]

print("Low Cardinality Features:", low_cardinality)
print("\nHigh Cardinality Features:", high_cardinality)

# 3. Frequency encoding for high-cardinality features
print("\n2. APPLYING FREQUENCY ENCODING")

for col in high_cardinality:

    frequency_map = (
        X_train_encoded[col]
        .value_counts(normalize=True)
        .to_dict()
    )

    X_train_encoded[col] = (
        X_train_encoded[col].map(frequency_map).fillna(0)
    )

    X_test_encoded[col] = (
        X_test_encoded[col].map(frequency_map).fillna(0)
    )

print("High-cardinality features encoded successfully.")

# 4. One-hot encoding for low-cardinality features
print("\n3. APPLYING ONE-HOT ENCODING")

X_train_encoded = pd.get_dummies(
    X_train_encoded,
    columns=low_cardinality,
    dtype=int
)

X_test_encoded = pd.get_dummies(
    X_test_encoded,
    columns=low_cardinality,
    dtype=int
)

# Align training and testing columns
X_train_encoded, X_test_encoded = X_train_encoded.align(
    X_test_encoded,
    join="left",
    axis=1,
    fill_value=0
)

# 5. Verify encoded datasets
print("\n4. ENCODING VALIDATION")

print("Training Dataset Shape:", X_train_encoded.shape)
print("Testing Dataset Shape:", X_test_encoded.shape)

print(
    "All Features Numerical:",
    all(
        pd.api.types.is_numeric_dtype(dtype)
        for dtype in X_train_encoded.dtypes
    )
)

print("\n5. SAMPLE ENCODED DATA")
display(X_train_encoded.head())

print("\nTASK 17 COMPLETED SUCCESSFULLY")
```

```python
# TASK 18: FEATURE SCALING USING STANDARDSCALER

print("=" * 60)
print("TASK 18: FEATURE SCALING USING STANDARDSCALER")
print("=" * 60)

from sklearn.preprocessing import StandardScaler

# 1. Identify numerical columns in encoded dataset
print("\n1. IDENTIFYING NUMERICAL FEATURES")

scaling_features = [
    col for col in numerical_features
    if col in X_train_encoded.columns
]

print("Features to Scale:", scaling_features)

# 2. Create StandardScaler
print("\n2. APPLYING STANDARD SCALER")

scaler = StandardScaler()

# 3. Fit scaler using training data only
X_train_encoded[scaling_features] = scaler.fit_transform(
    X_train_encoded[scaling_features]
)

# 4. Transform testing data
X_test_encoded[scaling_features] = scaler.transform(
    X_test_encoded[scaling_features]
)

print("Feature scaling completed successfully.")

# 5. Display scaled feature statistics
print("\n3. SCALED TRAINING FEATURE STATISTICS")

display(
    X_train_encoded[scaling_features].describe().round(3)
)

# 6. Verify training feature means
print("\n4. VERIFYING SCALED FEATURE MEANS")

display(
    X_train_encoded[scaling_features].mean().round(4)
)

# 7. Verify dataset dimensions
print("\n5. FINAL DATASET SHAPES")

print("Training Dataset:", X_train_encoded.shape)
print("Testing Dataset:", X_test_encoded.shape)

print("\nTASK 18 COMPLETED SUCCESSFULLY")
```

```python
# TASK 19: HANDLING CLASS IMBALANCE USING SMOTE

print("=" * 60)
print("TASK 19: HANDLING CLASS IMBALANCE USING SMOTE")
print("=" * 60)

from imblearn.over_sampling import SMOTE

# 1. Display original class distribution
print("\n1. ORIGINAL TRAINING CLASS DISTRIBUTION")

display(
    y_train.value_counts()
    .sort_index()
    .to_frame(name="Record Count")
)

# 2. Initialize SMOTE
print("\n2. APPLYING SMOTE")

smote = SMOTE(
    random_state=42,
    sampling_strategy="auto",
    k_neighbors=5
)

# 3. Balance training data
X_train_balanced, y_train_balanced = smote.fit_resample(
    X_train_encoded,
    y_train
)

print("SMOTE applied successfully.")

# 4. Display balanced class distribution
print("\n3. BALANCED TRAINING CLASS DISTRIBUTION")

display(
    pd.Series(y_train_balanced)
    .value_counts()
    .sort_index()
    .to_frame(name="Record Count")
)

# 5. Compare dataset dimensions
print("\n4. DATASET SHAPES")

print("Original Training Features:", X_train_encoded.shape)
print("Balanced Training Features:", X_train_balanced.shape)

print("Original Training Target:", y_train.shape)
print("Balanced Training Target:", y_train_balanced.shape)

# 6. Verify testing data remains unchanged
print("\n5. TESTING DATA VALIDATION")

print("Testing Features:", X_test_encoded.shape)
print("Testing Target:", y_test.shape)

print("\nTASK 19 COMPLETED SUCCESSFULLY")
```

```python
# TASK 20: SAVE PREPROCESSED DATASET

print("=" * 60)
print("TASK 20: SAVE PREPROCESSED DATASET")
print("=" * 60)

# 1. Create training dataset
print("\n1. CREATING TRAINING DATASET")

train_final = X_train_balanced.copy()
train_final["CLASS"] = y_train_balanced

print("Training Dataset Shape:", train_final.shape)

# 2. Create testing dataset
print("\n2. CREATING TESTING DATASET")

test_final = X_test_encoded.copy()
test_final["CLASS"] = y_test

print("Testing Dataset Shape:", test_final.shape)

# 3. Save datasets as CSV
print("\n3. SAVING DATASETS")

train_final.to_csv(
    "ClinVar_Training_Preprocessed.csv",
    index=False
)

test_final.to_csv(
    "ClinVar_Testing_Preprocessed.csv",
    index=False
)

print("Training dataset saved successfully.")
print("Testing dataset saved successfully.")

# 4. Verify saved datasets
print("\n4. VERIFYING SAVED DATASETS")

print("Training File Size:",
      round(os.path.getsize("ClinVar_Training_Preprocessed.csv") / (1024 * 1024), 2),
      "MB")

print("Testing File Size:",
      round(os.path.getsize("ClinVar_Testing_Preprocessed.csv") / (1024 * 1024), 2),
      "MB")

# 5. Display final dataset information
print("\n5. FINAL DATASET SUMMARY")

print("Training Dataset:", train_final.shape)
print("Testing Dataset:", test_final.shape)

print("\nTraining Class Distribution:")
display(train_final["CLASS"].value_counts().sort_index())

print("\nTesting Class Distribution:")
display(test_final["CLASS"].value_counts().sort_index())

print("\nTASK 20 COMPLETED SUCCESSFULLY")
```
